```python
#tuple operations
t1=(1,2,3,4)
t2=(5,6,'hello','hi')
#concatination
print(t1+t2)
#repitition
print(t1*3)
#membership
print(1 in t1)
print(6 not in t1)
#iteration
for i in t1:
  print(i,end=" ")
```

```python
#IN-BUILT functions
t3=(12,23,45,35,56)
print(len(t3))
print(sum(t3))
print(min(t3))
print(max(t3))
print(tuple(reversed(t3)))
print(tuple(sorted(t3)))
```

```python
#methods in tuple
t4=(2,3,4,6,7,8,9,9,9,9,9)
print(t4.index(3))
print(t4.count(9))
```

```python
a=('Rakshitha','Rohan','Rishitha')
b=(1,2,3)
print(tuple(zip(a,b))) #in tuple
print(list(zip(a,b))) #in list
```
