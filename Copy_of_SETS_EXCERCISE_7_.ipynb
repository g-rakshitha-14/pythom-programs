```python
#SET OPERATIONS
a={1,2,4,5,6}
b={6,7,8,9,'hello'}
print(a|b) #union
print(a&b) #intersection
print(a-b) #difference
print(a^b) #symmetric diference
```

```python
#SET OPERATIONS
a={1,2,3,4,5,'hi'}
b={1,2,3,4,7,0}
print("a= ",a)
print("b= ",b)
a.add(10)
print("add(): ",a)
a.update([11,12])
print("update(): ",a)
a.remove(10)
print("update(): ",a)
a.discard(11)
print("discard(): ",a)
print("union(): ",a)
print("intersection(): ",a)
print("difference(): ",a)
```

```python
#write a program to find how many numbers have duplicates
L1=[int(i) for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)>i:
    L2.append (i)
s=set(L2)
print(len(s))
```

```python
#wrilte a python program to print not repeated elements
L=[int(i) for i in input().split()]
l1=[]
for i in L:
  if L.count(i)==1:
    l1.append(i)
s=set(l1)
print(len(s))
```
