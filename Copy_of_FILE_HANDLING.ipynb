```python
with open("new.txt",'r') as f:
  data=f.read().split()
data_lower = [i.lower() for i in data]
sort_data = sorted(data_lower)
print(sort_data)
with open("f4.txt",'r') as f:
  for i in sort_data:
   f.write(i)
print("done")
```

```python
#python programme to compute the noof characters,words,and lines in a file
file = open("new.txt","r")
lines=file.readlines()
character=0
words=0
line_count=len(lines)
for line in lines:
  character+=len(line)
  words+=len(line.split())
print("no.of characters: ",character)
print("no.of words: ",words)
print("no.of lines: ",line_count)
file.closed
```
