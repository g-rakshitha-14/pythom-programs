```python
#DICTIONARY
d={"sno":35,"name":"Rakshitha","grade":10,"marks":[100,100]}
print(d)
```

```python
d=dict((("sno",2),("marks",100),("name","Ram"),("age",18)))
print(d)
```

```python
print(d["name"])
```

```python
(d["name"])="siri"
print(d["name"])
```

```python
d.pop("sno")
print(d)
```

```python
print(len(d))
```

```python
print(max(d))
```

```python
#dictionary comphrehention
d={i:i*i for i in range(1,6)}
print(d)
```

```python
#user defined dictionary
k=[i for i in input().split()]
v=[int(i)for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(d)
```
